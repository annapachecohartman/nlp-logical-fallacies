```python
# parse_comments.ipynb

import json
import pandas as pd
```

```python
with open('../dataJSON/train.json', 'r', encoding='utf-8') as f:
    data = json.load(f)
all_comments = []
for article in data:
    comments = article.get("comments", [])
    for comment in comments:
        all_comments.append({
            "comment": comment["comment"],
            "fallacy": comment["fallacy"]
        })
df_train = pd.DataFrame(all_comments)
```

```python
with open('../dataJSON/test.json', 'r', encoding='utf-8') as f:
    data = json.load(f)
all_comments = []
for article in data:
    comments = article.get("comments", [])
    for comment in comments:
        all_comments.append({
            "comment": comment["comment"],
            "fallacy": comment["fallacy"]
        })
df_test = pd.DataFrame(all_comments)
```

```python
with open('../dataJSON/dev.json', 'r', encoding='utf-8') as f:
    data = json.load(f)
all_comments = []
for article in data:
    comments = article.get("comments", [])
    for comment in comments:
        all_comments.append({
            "comment": comment["comment"],
            "fallacy": comment["fallacy"]
        })
df_dev = pd.DataFrame(all_comments)
```
